```python
#*Q13.* Fill missing values in Monthly_Income with the median income.
df["Monthly_Income"]=df["Monthly_Income"].fillna(df["Monthly_Income"]).median()
df[df["Monthly_Income"].isnull()]
```

```python
#*Q14.* Fill missing values in Credit_Score with the median credit score.
df["Credit_Score"]=df["Credit_Score"].fillna(df["Credit_Score"]).median()
df[df["Credit_Score"].isnull()]
```

```python
#*Q18.* Remove duplicate rows from the dataset.
df=df.drop_duplicates()
```

```python
# ----------------------------------------------------------
# TEXT CLEANING 
# ----------------------------------------------------------
```

```python
#*Q26.* Remove extra spaces from the Name column.
df["Name"]=df["Name"].str.strip()
```

```python
#*Q27.* Standardize the values in the Gender column so that M, Male, and MALE are represented consistently. Do the same for female values.
df["Gender"]=df["Gender"].replace({"M":"Male", "MALE":"Male","male":"Male","FEMALE":"Female","female":"Female","F":"Female"})
df["Gender"]=df["Gender"].str.strip()
df["Gender"].unique()
```

```python
#*Q28.* Find the unique values in the City column.
df["City"].unique()
```

```python
#Q30.* Correct spelling mistakes in city names where appropriate.
df["City"]=df["City"].str.strip()
df["City"]=df["City"].replace({"New Delhi":"Delhi","Bombay":"Mumbai","Bangalore":"Bengaluru","Dehli":"Delhi"})
df["City"].unique()
```

```python
# ----------------------------------------------------------
# INVALID VALUES DETECTION
# ----------------------------------------------------------
```

```python
#*Q31.* Find customers whose age is less than 18 or greater than 100.
df[(df["Age"]<18) | (df["Age"] >100)]
```

```python
#*Q32.* Find customers with negative monthly income.
df[df["Monthly_Income"] < 0]
```

```python
#Q33.* Find customers with negative loan amounts.
df[df["Loan_Amount"]<0]
```

```python
#*Q34.* Find customers with monthly income greater than ₹1,00,000.
df[df["Monthly_Income"]>100000]
```
